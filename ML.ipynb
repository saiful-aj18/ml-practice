```python
import pandas as pd
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
df = pd.DataFrame({
    'age': [25, 23, 30, 22],
    'city': ['Dhaka', 'Chattogram', 'Khulna', 'Sylhet'],
    'salary': [40000, 30000, 25000, 65000]
})
# 1. Fill missing values with the mean age
df['age'] = SimpleImputer(strategy='mean').fit_transform(df[['age']])
# 2. Convert city names into 0/1 columns (one-hot encoding)
city_encoded = OneHotEncoder(sparse_output=False).fit_transform(df[['city']])
# 3. Bring salary onto the same scale (standardization)
df['salary'] = StandardScaler().fit_transform(df[['salary']])
print(df)
print(city_encoded)
```

## 4. Training a Simple ML Model

Now we will train a linear regression model to predict salary using the processed data (features).

```python
import numpy as np
from sklearn.linear_model import LinearRegression
# Build the features: combine age with the encoded city columns
X = np.hstack((df[['age']].values, city_encoded))
y = df['salary'].values  # Target variable
# Create and fit the model
model = LinearRegression()
model.fit(X, y)
# Make predictions
predictions = model.predict(X)
# Compare actual vs predicted salaries
result_df = pd.DataFrame({
    'Actual Salary (Scaled)': y,
    'Predicted Salary (Scaled)': predictions
})
display(result_df)
```

## 5. Visualization of Actual vs Predicted Salaries

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set the plot style
sns.set_theme(style="whitegrid")

# Melt the DataFrame for easier plotting
df_melted = result_df.reset_index().melt(id_vars='index', value_vars=['Actual Salary (Scaled)', 'Predicted Salary (Scaled)'], 
                                         var_name='Type', value_name='Salary')

plt.figure(figsize=(10, 6))
sns.barplot(x='index', y='Salary', hue='Type', data=df_melted)

plt.title('Comparison: Actual vs Predicted Salaries (Scaled)')
plt.xlabel('Data Point Index')
plt.ylabel('Salary (Scaled Value)')
plt.xticks(ticks=range(4), labels=[f'Person {i+1}' for i in range(4)])
plt.legend(title='Category')
plt.show()
```
