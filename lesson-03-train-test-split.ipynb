```python
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

# Lesson 2 data
df = pd.DataFrame({
    'age': [25, 23, 30, 22],
    'city': ['Dhaka', 'Chattogram', 'Khulna', 'Sylhet'],
    'salary': [40000, 30000, 25000, 65000]
})

df['age'] = SimpleImputer(strategy='mean').fit_transform(df[['age']])
city_encoded = OneHotEncoder(sparse_output=False).fit_transform(df[['city']])
df['salary'] = StandardScaler().fit_transform(df[['salary']])

X = pd.concat([df[['age', 'salary']], pd.DataFrame(city_encoded)], axis=1)
y = np.array([65000, 70000, 50000, 45000])

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)

model = LinearRegression()
model.fit(X_train, y_train)

pred = model.predict(X_test)
print("Test error:", mean_squared_error(y_test, pred))
print("Actual:", y_test)
print("Predicted:", pred)
```
