```python
import pandas as pd
from sklearn.linear_model import LinearRegression

df = pd.DataFrame({
    'age': [25, 23, 30, 22],
    'salary': [40000, 30000, 25000, 65000]
})

X = df[['age']]     # feature: age only
y = df['salary']    # target: salary

model = LinearRegression()
model.fit(X, y)

print("Slope (m):", model.coef_[0])
print("Intercept (b):", model.intercept_)
print("28 বছর বয়সে salary:", model.predict([[28]]))
```
