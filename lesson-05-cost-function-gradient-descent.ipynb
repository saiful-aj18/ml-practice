```python
import numpy as np

X = np.array([1, 2, 3, 4])
y = np.array([2, 4, 6, 8])

m, b = 0.0, 0.0
lr = 0.0001  # learning rate

for i in range(1000):
    y_pred = m * X + b
    error = y_pred - y
    m -= lr * (2/len(X)) * np.sum(error * X)
    b -= lr * (2/len(X)) * np.sum(error)

print("m =", m, " b =", b)          # ~2.0, ~0.0
print("MSE =", np.mean((y - (m*X + b))**2))  # ~0
```
