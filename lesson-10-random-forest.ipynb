# Lesson 10: Random Forest

One Decision Tree overfits easily (Lesson 6). Random Forest fixes this with many trees: each tree trains on a random bootstrap sample (bagging) and looks at a random subset of features at each split, so the trees differ from each other. For classification the final answer is the majority vote of all trees. `n_estimators` = number of trees (100 is a common default), and `feature_importances_` shows which feature mattered most.

```python
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import cross_val_score

X, y = load_iris(return_X_y=True)

tree = DecisionTreeClassifier(random_state=42)
forest = RandomForestClassifier(n_estimators=100, random_state=42)

print("Decision Tree:", cross_val_score(tree, X, y, cv=5).mean())   # 0.953
print("Random Forest:", cross_val_score(forest, X, y, cv=5).mean()) # 0.967

forest.fit(X, y)
print(forest.feature_importances_)
```

## Practice

Try `n_estimators=10`, `100`, `500` and see how the score changes. Then use `feature_importances_` to find the most important Iris feature.

```python
# Practice: how does the score change with the number of trees?
for n in [10, 100, 500]:
    rf = RandomForestClassifier(n_estimators=n, random_state=42)
    print(f"n_estimators={n}:", round(cross_val_score(rf, X, y, cv=5).mean(), 3))

# Which Iris feature matters most?
rf100 = RandomForestClassifier(n_estimators=100, random_state=42).fit(X, y)
names = load_iris().feature_names
print("Most important feature:", names[rf100.feature_importances_.argmax()])
print("Importances:", dict(zip(names, rf100.feature_importances_.round(3))))
```
